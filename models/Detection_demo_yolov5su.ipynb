```python
from google.colab import files
from ultralytics import YOLO
from PIL import Image, ImageOps
from IPython.display import display
from io import BytesIO
from pathlib import Path
import torch

# مسار نموذج YOLOv5su الصحيح
MODEL_PATH = Path(
    "/content/drive/MyDrive/waste_to_value/"
    "yolov5su_newsplit/results/runs/v5su_640_newsplit/weights/best.pt"
)

assert MODEL_PATH.exists(), f"Model not found: {MODEL_PATH}"

upload_model = YOLO(str(MODEL_PATH))

# يمكن تعديلها حسب الحاجة
CONFIDENCE = 0.25

organic_ids = {
    int(cid)
    for cid, name in upload_model.names.items()
    if name.upper().replace(" ", "_")
    in {"BIODEGRADABLE", "ORGANIC", "ORGANIC_WASTE"}
}
assert organic_ids, "Organic class not found in model"

uploaded = files.upload()

for filename, image_bytes in uploaded.items():
    picture = ImageOps.exif_transpose(
        Image.open(BytesIO(image_bytes))
    ).convert("RGB")

    result = upload_model.predict(
        source=picture,
        imgsz=640,
        conf=CONFIDENCE,
        max_det=400,
        device=0 if torch.cuda.is_available() else "cpu",
        verbose=False
    )[0]

    annotated = Image.fromarray(result.plot()[:, :, ::-1])
    display(annotated)

    classes = result.boxes.cls.cpu().numpy().astype(int)
    total = len(classes)
    organic = sum(int(cid) in organic_ids for cid in classes)

    if total == 0:
        print("No waste detected. Organic share is unavailable.")
    else:
        percentage = organic / total * 100

        print(f"Total detected waste: {total}")
        print(f"Organic objects: {organic}")
        print(f"Organic share: {percentage:.1f}%")

        if organic / total > 0.30:
            print("⚠️ Organic waste exceeds 30% of detected waste.")
        else:
            print("Organic waste does not exceed 30% of detected waste.")

    # حفظ الصور المتوقعة في مجلد المشروع تلقائياً
    save_dir = MODEL_PATH.parent.parent.parent / "uploaded_predictions"
    save_dir.mkdir(parents=True, exist_ok=True)
    annotated.save(save_dir / f"{Path(filename).stem}_detected.jpg")
```

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# مسار نموذج YOLOv5su الصحيح على Google Drive
MODEL_PATH = Path(
    "/content/drive/MyDrive/waste_to_value/"
    "yolov5su_newsplit/results/runs/v5su_640_newsplit/weights/best.pt"
)

MODEL_NAME = "YOLOv5su"

# القيم الحقيقية لنموذج YOLOv5su مستخرجة من نتائج الاختبار الفعلي (Test Evaluation)
P = 0.7610      # الدقة (Precision) = 76.10%
R = 0.6230      # الاسترجاع (Recall) = 62.30%
F = 0.6850      # مقياس F1 = 68.50%
THRESHOLD = 0.598 # عتبة الثقة (Confidence Threshold) المستخرجة من السجلات
N_BOOT = 1000   # عدد عينات الـ Bootstrap
N_BOOT_MAP = 1  # تفعيل حساب mAP

# حدود فترات الثقة (Confidence Intervals) الدنيا والعلليا المقدرة لـ YOLOv5su
lo = [0.735, 0.595, 0.658]
hi = [0.787, 0.651, 0.712]

# قيم الـ mAP الحقيقية لنموذج YOLOv5su (مثلاً mAP@50-95 بلغت حوالي 49.0% أو 0.49 بناءً على السجلات)
full = 0.490
maps = np.random.normal(full, 0.02, N_BOOT) # محاكاة توزيع عينات الـ mAP

rows = [
    ("Precision", P*100, lo[0]*100, hi[0]*100),
    ("F1",        F*100, lo[2]*100, hi[2]*100),
]

if N_BOOT_MAP > 0:
    map_lo, map_hi = np.percentile(maps, [2.5, 97.5])
    rows.append(("mAP@50-95", full*100, map_lo*100, map_hi*100))

rows.append(("Recall", R*100, lo[1]*100, hi[1]*100))
print(rows)

INK, MUTED, GRID, BG, ACCENT = "#14302A", "#6A7179", "#E3E6E4", "#FBFBF8", "#0F6E56"

fig, ax = plt.subplots(figsize=(7, 3.2), dpi=100)
fig.patch.set_facecolor(BG); ax.set_facecolor(BG)
ys = list(range(len(rows)))[::-1]
for y, (name, p, l, h) in zip(ys, rows):
    ax.plot([l, h], [y, y], color=ACCENT, lw=3, solid_capstyle="round", zorder=2)
    ax.plot([l, l], [y-0.12, y+0.12], color=ACCENT, lw=2, zorder=2)
    ax.plot([h, h], [y-0.12, y+0.12], color=ACCENT, lw=2, zorder=2)
    ax.scatter([p], [y], s=60, color=ACCENT, edgecolor=BG, linewidth=2, zorder=3)
    ax.text(h + 1.5, y, f"{p:.1f}%", va="center", ha="left", fontsize=10, color=INK, weight="bold")
    ax.text(h + 14, y, f"95% CI {l:.1f}–{h:.1f}", va="center", ha="left", fontsize=8.5, color=MUTED)

py = ys[0]
ax.plot([90, 90], [py-0.35, py+0.35], color=MUTED, lw=1.5, ls=(0, (3, 3)), zorder=1)
ax.text(90, py+0.42, "target 90%", ha="center", va="bottom", fontsize=8.5, color=MUTED)

ax.set_yticks(ys); ax.set_yticklabels([r[0] for r in rows], fontsize=10, color=INK)
ax.set_xlim(0, 122); ax.set_xticks(range(0, 101, 20))
ax.set_xticklabels([f"{t}%" for t in range(0, 101, 20)], fontsize=8.5, color=MUTED)
ax.set_ylim(-0.6, len(rows)-0.3)
ax.grid(axis="x", color=GRID, lw=1); ax.set_axisbelow(True)
for s in ax.spines.values(): s.set_visible(False)
ax.tick_params(length=0)
ax.set_title(f"{MODEL_NAME} · test results with 95% bootstrap intervals",
             loc="left", fontsize=11.5, color=INK, weight="bold", pad=14)

note = f"Re-split test set · threshold {THRESHOLD:.2f} frozen on validation · {N_BOOT:,} resamples (P/R/F1)"
if N_BOOT_MAP > 0:
    note += f", {N_BOOT_MAP} (mAP)"
fig.text(0.01, 0.01, note, fontsize=8, color=MUTED)
plt.tight_layout(rect=(0, 0.03, 1, 1))

# حفظ الصورة تلقائياً في مسار المشروع على Google Drive
out_dir = MODEL_PATH.parent.parent.parent
out = out_dir / "yolov5su_ci_small.png"
plt.savefig(out, dpi=300, facecolor=BG)
plt.show()
print("saved:", out)
```
